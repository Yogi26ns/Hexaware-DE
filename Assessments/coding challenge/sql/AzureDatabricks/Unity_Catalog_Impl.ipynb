##Project Implementation

This section demonstrates step by step how Unity Catalog is implemented in Databricks for organizing and securing data assets.

####Create a Catalog

A Catalog is the top-level container in Unity Catalog.

```python
%sql
-- Create a new catalog
CREATE CATALOG demo_catalog
COMMENT "This catalog is created for Unity Catalog mini project demonstration";
```

####Create a Schema (Database) inside Catalog

Schemas (databases) group related tables together.

```python
%sql
-- Create schema under catalog
CREATE SCHEMA demo_catalog.sales_data
COMMENT "Schema for sales-related datasets";
```

####Create a Delta Table
Now create a Delta table inside the schema.

```python
%sql
-- Create a managed Delta table
CREATE TABLE demo_catalog.sales_data.transactions (
    id INT,
    product STRING,
    quantity INT,
    price DOUBLE,
    sale_date DATE
)
USING DELTA
COMMENT "Sales transactions data table";
```

####Insert Sample Data

```python
%sql
INSERT INTO demo_catalog.sales_data.transactions
VALUES 
(1, 'Laptop', 2, 75000.0, '2023-01-12'),
(2, 'Mobile', 5, 15000.0, '2023-02-01'),
(3, 'Tablet', 1, 30000.0, '2023-03-05');
```

####Query the Table

```python
%sql
SELECT * FROM demo_catalog.sales_data.transactions;
```

####Apply Permissions
Unity Catalog allows fine-grained access control.

```python
%sql
-- Grant SELECT access on the table to a specific user
GRANT SELECT ON TABLE demo_catalog.sales_data.transactions TO `nsyogiii@gmail.com`;

-- Revoke permissions if needed
REVOKE SELECT ON TABLE demo_catalog.sales_data.transactions FROM `nsyogiii@gmail.com`;
```
